```python
import pandas as pd 

import kagglehub

# Download latest version
path = kagglehub.dataset_download("jasonqhiu/sp500")

print("Path to dataset files:", path)

df = pd.read_csv(f"{path}/GSPC.csv") #más rápido
```

Limpieza de datos

```python
tablaBooleanaNan = df.isna() #Tabla si hay NaN o no
tablaBooleanaNan.sum() #sumatorio para dataframes
df.dropna() #Adios datos vacíos
df.fillna("Hola pana") #los NaN se le añade el valor puesto 
```

```python
df.duplicated() #estan las filas duplicadas?
df.drop_duplicates() #adios filas duplicadas
```

```python
df['Date'].str.lower() #pasar a minúscula la columna entera
df['Date'].str.lower() #pasar a mayuscula la columna entera
df['Date'].str.strip() #quitar los espacios al principio y final
df['Date'].str.replace("/", ".") #cambiar los / por .
```

```python
df.rename(columns={'Date': "Fecha_bro"}, inplace=True) #OJO: nombre viejo, nombre nuevo 
#OJO: inplace: reemplazo el nombre de la original o no?
df.replace (0, 1) #elementos se ponen como lo diga el tipo de dato
df["Fecha_bro"] = df["Fecha_bro"].astype(str)
```

Filtrar información

```python
a = df[df["Close"] > 100] #Se queda solo con los valores que cumplen o no
#Dentro debe haber una condición booleana, para el y -> & y para el o -> |
b = df[(df["Close"] > 100) & (df["Close"] < 200)] #Se queda solo con los valores que cumplen o no

datos = [100, 200, 300]
c = df[df["Close"].isin(datos)] #Se queda solo con los valores que están en la lista
d= df[df["Fecha_bro"].str.contains("2020", case=False, na=False)] #Se queda solo con los valores que contienen la cadena "2020" en la columna "Fecha_bro"
#Case indica si importa mayuscula o minuscula, na indica si se pueden los datos NaN

#OJO: mejor utilizar .query, mas limpio
p1 = 100
p2 = 200
df.query("Close > @p1 and Close < @p2") 
#Directamente se pone el nombre de las columnas y los operadores lógicos
#Se pone and y or en vez de & y |
#Para las variables, se pone @variable
u = df.query("Close in @p1")
v = df.query("Close not in @p1")
#eN VEZ DE USAR EL .isin() se puede usar el .query()
#OJO: si hay un espacio en medio, se pone `nombre de la columna`, son comillas invertidas
#OJO: cuidado si cambian muchos los valores
```

Operaciones avanzadas

```python
c = pd.concat([a, b]) #Concatenar dos dataframes
print(c.head()) #Ver la cabeza

d = pd.merge(a, b, on="Fecha_bro", how="inner") #Datos en común que junta para crear un nuevo dataframe
print(d.head()) #Ver la cabeza
#Inner: solo intersección común ambos, left: todo izquierda, right: todo derecha, outer: todo los datos
```

```python
d = a.groupby("Fecha_bro").sum() #Ojo, se agrupa, se selecciona la columna y se palica una operación
```
